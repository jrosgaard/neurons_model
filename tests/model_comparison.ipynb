```python
# Import NEURON and check its version
import neuron; print(neuron.__version__)
from neuron import h, rxd
from neuron.units import ms, mV
```

```python
# Create a simple two-compartment model with Hodgkin-Huxley dynamics in soma1 and passive properties in soma2.

# Create soma1 and insert Hodgkin-Huxley channels
soma1 = h.Section(name="soma1")
soma1.insert("hh")

# Create soma2 and insert passive properties
soma2 = h.Section(name="soma2")
soma2.insert("pas")

h.topology()
```

```python
# Clamp the soma1 and soma2 with a 1 MΩ resistance
iclamp1 = h.IClamp(soma1(0.5))
iclamp2 = h.IClamp(soma2(0.5))

# Set the membrane potential vectors of soma1 and soma2
v1 = h.Vector().record(soma1(0.5)._ref_v)
v2 = h.Vector().record(soma2(0.5)._ref_v)

# Time stamp vector
t = h.Vector().record(h._ref_t)

# Load the standard run library and initialize the simulation
h.load_file("stdrun.hoc")
h.finitialize(-65 * mV)

# Run the simulation for 40 ms
h.continuerun(40 * ms)
```

```python
# Plot the membrane potential over time for both soma1 and soma2

plt.figure(figsize=(10, 8))
#plt.subplot(2,1,1)

plt.plot(t, v1, label='soma1 (hh)', color='blue', linestyle='--',)
plt.plot(t, v2, label='soma2 (pas)', color='red', linestyle='-',)

plt.title('Membrane Potential of Two Sections')
plt.xlabel('Time (ms)')
plt.ylabel('Membrane Potential (mV)')

plt.grid()
plt.legend('soma1 (hh)', 'soma2 (pas)')

plt.tight_layout()
plt.show()
```
